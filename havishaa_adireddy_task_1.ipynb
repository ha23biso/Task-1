1. Safecracker: Palindrome and Digit Sum


```python
rd,og=0,0
```

```python
def can_crack(n):
    og = n
    rev = 0
    sum = 0
    while n > 0:
        rd = n % 10
        sum = sum + rd
        rev = rev * 10 + rd
        n = n // 10
    if rev == og and og % sum == 0:
        return True
    
    else:
        return False
```

```python
num = int(input("Enter a number: "))
can_crack(num)
    
```

2. Most frequent word program

```python
f = open("text.txt","w")
f.write("The cat sat.\nThe CAT ran!\nThe dog barked, but the cat slept.")
f.close()

```

```python
d = {}
f = open("text.txt","r")
x = f.readlines()
for i in x:
    for j in i.split():
        j = j.lower()
        if j in d:
            d[j] += 1
        else:
            d[j] = 1
f.close()
k = max(d, key=d.get)
print(f"Most frequent word:'{k}'({d[k]} times)")
```

```python
w = ""
for k, v in d.items():
    if v > 1 and len(k) > len(w):
        w = k
print(f"Longest repeated word:'{w}'")
```

3. Staircase Climbing: Recursion and Memoization

```python
import time
def count_ways_recursive(n):
    if n < 0:
        return 0
    if n == 0:
        return 1
    
    return (count_ways_recursive(n - 1) + count_ways_recursive(n - 2) + count_ways_recursive(n - 3))
```

```python
n = 30
start = time.time()
ways_rec = count_ways_recursive(n)
time_rec = time.time() - start
print(f"Plain Recursion: {ways_rec} ways | Time taken: {time_rec:.5f} seconds")
```

```python

def count_ways_memo(n, x=None):
    if x is None:
        x = {}
    
    if n < 0:
        return 0
    if n == 0:
        return 1

    if n in x:
        return x[n]
    
    x[n] = (count_ways_memo(n - 1, x) + count_ways_memo(n - 2, x) + count_ways_memo(n - 3, x))
    
    return x[n]
```

```python
start = time.time()
ways_memo = count_ways_memo(n)
time_memo = time.time() - start
print(f"Memoized Recursion: {ways_memo} ways | Time taken: {time_memo:.5f} seconds")
```

4. Collatz conjecture

```python
def collatz_steps(n):
    l = []
    l.append(n)
    a = 0
    while n != 1:
        if n % 2 ==0:
            k = n/2
        else:
            k = 3*n + 1
        l.append(k)
        a+=1
        n = k
    print(f"{a} steps")
```

```python
for i in range(1,10001):
    n = i
    collatz_steps(n)
```

5. Library management system

```python
class Item:
    def __init__(self, title):
        self._title = title
        self._is_borrowed = False

    def get_title(self):
        return self._title

    def is_borrowed(self):
        return self._is_borrowed

    def borrow(self):
        if not self._is_borrowed:
            self._is_borrowed = True
            return True
        return False

    def return_item(self):
        if self._is_borrowed:
            self._is_borrowed = False
            return True
        return False

    def __str__(self):
        return f"Item: '{self._title}'"
```

```python
class Book(Item):
    def __init__(self, title, author, pages):
        super().__init__(title)
        self.author = author
        self.pages = pages
        
    def __str__(self):
        return f"Book: '{self.get_title()}' by {self.author} ({self.pages} pages)"
```

```python

class DVD(Item):
    def __init__(self, title, duration_minutes):
        super().__init__(title)
        self.duration_minutes = duration_minutes

    def __str__(self):
        return f"DVD: '{self.get_title()}' ({self.duration_minutes} mins)"
```

```python
class Library:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def borrow_item(self, title):
        for item in self.items:
            if item.get_title().lower() == title.lower():
                if item.is_borrowed():
                    print(f"'{title}' already borrowed.")
                else:
                    item.borrow()
                    print(f"u borrowed '{title}'.")
                return
        print(f"'{title}' not in library.")

    def show_available(self):
        print("Available Items:")
        for item in self.items:
            if not item.is_borrowed():
                print(item)
```

```python
lib = Library()
lib.add_item(Book("Atomic Habits", "James Clear", 320))
lib.add_item(DVD("Inception", 148))
lib.borrow_item("Inception")
print()
lib.show_available()
```

6. Circle Class


```python
import math as m
class Circle:
    def __init__(self, a, b, r):
        self.a = a
        self.b = b
        self.r = r

    def Area(self):
        return m.pi * pow(self.r, 2)

    def Perimeter(self):
        return 2 * m.pi * self.r

    def testBelongs(self, x, y):
        return pow((x - self.a), 2) + pow((y - self.b), 2) <= pow(self.r, 2)
```

```python

c = Circle(0, 0, 7)
print(f"Center O: ({c.a}, {c.b})")
print(f"Radius r: {c.r}")
print(f"Area: {c.Area():.2f}")
print(f"Perimeter: {c.Perimeter():.2f}")
```

7. Computation Class

```python
class Computation:
    def __init__(self):
        pass

    def factorial(self, n):
        prod = 1
        for i in range(1, n + 1):
            prod *= i
        return prod

    def sum(self, n):
        tot = 0
        tot = n * (n + 1) / 2
        return tot

    def testPrim(self, n):
        if n < 2:
            return False
        for i in range(2, n):
            if n % i == 0:
                return False
        return True

    def testPrims(self, a, b):
        while b != 0:
            a, b = b, a % b
        if a==1:
            return True
        else:
            return False

k = Computation()

print("7!:", k.factorial(7))
print("Sum of first 3 integers:", k.sum(3))
print("2 prime?:", k.testPrim(2))
print("coprime?:", k.testPrims(8, 15))
```

8. Inheritance

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def displayPerson(self):
        print(f"name:{self.name},age:{self.age}")
```

```python
class Student(Person):
    def __init__(self, name, age, rollno, course):
        super().__init__(name, age)
        self.rollno = rollno
        self.course = course

    def displayStudent(self):
        print(f"rollno.:{self.rollno},course:{self.course}")
```

```python
class Teacher(Person):
    def __init__(self, name, age, employeeId, subject):
        super().__init__(name, age)
        self.employeeId = employeeId
        self.subject = subject

    def displayTeacher(self):
        print(f"employeeID:{self.employeeId},subject:{self.subject}")
```

```python
stu = Student("ram", 20, "22", "CSE")
teach = Teacher("ravi", 45, "mit_22", "math")

stu.displayPerson()
stu.displayStudent()
teach.displayPerson()
teach.displayTeacher()
```

10. Hill Number Check

```python
def is_hill_number(n):
    l = list(n)
    m = max(l)
    for i in range(len(l)):
        if l[i] == m:
            index = i
            break
    for i in range(index):
        if l[i] >= l[i + 1]:
            return False
    for i in range(index, len(l) - 1):
        if l[i] <= l[i + 1]:
            return False
    return True
```

```python
print(is_hill_number("12321"))
print(is_hill_number("12325"))
```

2. Singly Linked List

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert_end(self, x):
        a = Node(x)
        if not self.head:
            self.head = a
            return 
        b = self.head
        while b.next:
            b = b.next
        b.next = a

    def delete_end(self):
        if not self.head:
            print("list empty")
            return
        if not self.head.next:
            self.head = None
            return
        b = self.head
        while b.next.next:
            b = b.next
        b.next = None

    def display(self):
        b = self.head
        res = []
        while b:
            res.append(str(b.val))
            b = b.next
        print("  ".join(res) if res else "List empty")
```

```python
ll = LinkedList()
while True:
    ch = input("\n1-insert, 2-delete, 3-show, 4-exit")
    if ch == "1":
        ll.insert_end(input("value: "))
    elif ch == "2":
        ll.delete_end()
    elif ch == "3":
        ll.display()
    elif ch == "4":
        break
```
